```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 7, 9, 11, 13], dtype=float)
```

```python
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Sample Dataset")
plt.show()
```

```python
m = 0
b = 0
learning_rate = 0.01
epochs = 1000
n = len(X)

cost_history = []
```

```python
for i in range(epochs):
    y_pred = m * X + b

    # Calculate error (Mean Squared Error)
    cost = np.mean((y_pred - y) ** 2)
    cost_history.append(cost)

    # Calculate gradients
    dm = (2/n) * np.sum(X * (y_pred - y))
    db = (2/n) * np.sum(y_pred - y)
```

```python
plt.scatter(X, y, label="Actual Data")
plt.plot(X, m * X + b, color="red", label="Predicted Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent - Best Fit Line")
plt.legend()
plt.show()
```

```python
plt.plot(cost_history)
plt.xlabel("Epochs")
plt.ylabel("Cost (MSE)")
plt.title("Error Reduction During Gradient Descent")
plt.grid(True)
plt.show()
```
